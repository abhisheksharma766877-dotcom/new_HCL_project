```python
import pandas as pd
import numpy as np

# 1. Create a mock dataset for Movies
movies_data = {
    'movie_id': [1, 2, 3, 4, 5],
    'title': ['The Matrix', 'Inception', 'Interstellar', 'The Dark Knight', 'Avengers'],
    'genre': ['Sci-Fi', 'Sci-Fi', 'Sci-Fi', 'Action', 'Action']
}
movies = pd.DataFrame(movies_data)
display(movies)
```

```python
# 2. Create a mock dataset for User Ratings
ratings_data = {
    'user_id': [101, 101, 101, 102, 102, 103, 103, 103, 104, 104],
    'movie_id': [1, 2, 3, 1, 4, 2, 3, 5, 4, 5],
    'rating': [5, 4, 5, 4, 5, 5, 4, 4, 5, 3]
}
ratings = pd.DataFrame(ratings_data)
display(ratings)
```

```python
# 3. Merge Datasets
df = pd.merge(ratings, movies, on='movie_id')
display(df.head())
```

```python
# 4. Calculate Average Rating and Number of Ratings per Movie
movie_stats = df.groupby('title')['rating'].agg(['mean', 'count']).reset_index()
movie_stats.columns = ['Title', 'Average Rating', 'Number of Ratings']

# Sort to find the most popular and highest-rated movies
popular_movies = movie_stats.sort_values(by=['Number of Ratings', 'Average Rating'], ascending=[False, False])
display(popular_movies)
```

```python
# 5. Simple Recommendation Function (Based on Genre and Popularity)
def recommend_movies(genre, min_ratings=2):
    # Filter movies by genre
    genre_movies = movies[movies['genre'] == genre]

    # Merge with stats
    recs = pd.merge(genre_movies, movie_stats, left_on='title', right_on='Title')

    # Filter by minimum ratings and sort by average rating
    recs = recs[recs['Number of Ratings'] >= min_ratings]
    recs = recs.sort_values(by='Average Rating', ascending=False)

    return recs[['title', 'Average Rating', 'Number of Ratings']]

# Example: Recommend Sci-Fi movies
print("Sci-Fi Recommendations:")
display(recommend_movies('Sci-Fi', min_ratings=2))
```
